```python
#positional argument
def add_two_numbers(a,b):
  c = a + b
  return c
```

```python
add_two_numbers(1,2)
```

```python
a = 1
b = 5
c = add_two_numbers(a,b)

if c%2 == 0:
   print ('even')
else:
  print ('odd')
```

```python
#keyword arguments
def add_numbers(a,b):
  print (a+b)
```

```python
add_numbers(a=2,b=5)
```

```python
#default arguments
def register(name,mobileno,skils,location = 'hyderabad'):
    print(name,mobileno,skils,location)
print("registered")
```
